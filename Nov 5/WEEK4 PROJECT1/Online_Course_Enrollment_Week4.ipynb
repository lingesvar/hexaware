```python
import pandas as pd
from datetime import datetime
students = pd.read_csv("students.csv")
courses = pd.read_csv("courses.csv")
enrollments = pd.read_csv("enrollments.csv")
progress = pd.read_csv("progress.csv")
data = enrollments.merge(
    students, on="student_id"
)

data = data.merge(
    courses, on="course_id"
)

data = data.merge(
    progress, on="enrollment_id"
)
low_progress = data[data["completion"] < 50]
report = low_progress[
    ["student_name", "course_name", "completion", "status"]
]

print("Weekly Student Progress Report")
print(report)
report.to_csv(
    "weekly_progress_report.csv",
    index=False
)
with open("weekly_report_log.txt", "a") as file:
    file.write(
        f"Report generated: {datetime.now()}\n"
    )
    file.write(
        f"Students below 50% progress: {len(report)}\n"
    )

print("Weekly report generated successfully!")
```
