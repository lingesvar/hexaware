```python
import pandas as pd
from datetime import datetime
df = pd.read_csv("stockmovements.csv")
df["quantity"] = df["quantity"].astype(int)
stock_summary = df.groupby(
    ["product_id", "product_name", "warehouse_id"]
)["quantity"].sum().reset_index()
low_stock = stock_summary[
    stock_summary["quantity"] < 10
].copy()
low_stock = low_stock.rename(
    columns={"quantity": "current_stock"}
)
low_stock.to_csv(
    "daily_reorder_report.csv",
    index=False
)
with open("daily_stock_log.txt", "a") as file:
    file.write(
        f"Stock check completed: {datetime.now()}\n"
    )
    file.write(
        f"Products needing reorder: {len(low_stock)}\n"
    )

print("Daily Stock Check Completed!")
print("Products needing reorder:")
print(low_stock)

print("Reorder report generated successfully!")
```
